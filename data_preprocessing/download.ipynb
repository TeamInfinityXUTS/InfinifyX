### Segment Download

```python
import kagglehub
import os
import shutil


target_base_dir = "."


os.makedirs(target_base_dir, exist_ok=True)

# 2. 下载数据集（这会先下载到默认缓存路径）
print("Downloading...")
cache_path = kagglehub.dataset_download("solesensei/solesensei_bdd100k")

# 3. 确定数据集文件夹名称并将移动到目标位置
dataset_name = os.path.basename(cache_path)
final_destination = os.path.join(target_base_dir, dataset_name)

if os.path.exists(final_destination):
    print(f"目标路径 {final_destination} 已存在，正在清理以准备更新...")
    shutil.rmtree(final_destination)

# 移动文件夹
shutil.move(cache_path, target_base_dir)

print("-" * 30)
print(f"下载并移动完成！")
print(f"数据集当前位置: {os.path.abspath(final_destination)}")
```
