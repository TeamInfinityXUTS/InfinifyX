```python
import os

# Secure Configuration Loading
config_path = os.path.abspath(os.path.join(os.getcwd(), '..', 'clearml.conf'))
if os.path.exists(config_path):
    os.environ['CLEARML_CONFIG_FILE'] = config_path
    print(f"[*] Loaded secure ClearML configuration from: {config_path}")
else:
    print(f"[!] Warning: Local clearml.conf not found at {config_path}.")

```
