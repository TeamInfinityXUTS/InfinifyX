```python
import os

# Secure Configuration Loading
config_path = os.path.abspath(os.path.join(os.getcwd(), '..', 'clearml.conf'))
if os.path.exists(config_path):
    os.environ['CLEARML_CONFIG_FILE'] = config_path
    print(f"[*] Loaded secure ClearML configuration from: {config_path}")
else:
    print(f"[!] Warning: Local clearml.conf not found at {config_path}.")

```

```python
from clearml import Dataset

# 1. 创建一个新的数据集对象
# dataset_project: ClearML 界面上的项目名称
# dataset_name: 数据集的名称
dataset = Dataset.create(
    dataset_project='InfinifyX', 
    dataset_name='bdd100k_augmented'
)

# 2. 指定你要上传的本地文件夹路径
# 按照你的描述，路径是 'InfinifyX/bdd100k_tiny'
dataset.add_files(path='bdd100k_augmented')

# 3. 开始上传到 ClearML 文件服务器
dataset.upload()

dataset.finalize()

print("Dataset Upload")
```

### Upload Segmentation Dataset

```python
from clearml import Dataset

# 1. Create a new Dataset object
# Project name: 'InfinifyX', Dataset name: 'bdd100k_seg'
dataset = Dataset.create(
    dataset_project='InfinityX', 
    dataset_name='bdd100k_seg'
)

# 2. Specify the local folder path to upload
# This path contains color_labels, images, and labels folders
local_data_path = '/home/sagemaker-user/InfinifyX/data_preprocessing/dataset_seg/bdd100k_seg/bdd100k/seg'

# Verify if the path exists before adding files
if os.path.exists(local_data_path):
    print(f"Path confirmed: {local_data_path}")
    # Add all files from the directory to the ClearML dataset
    dataset.add_files(path=local_data_path)
else:
    print(f"ERROR: Could not find path {local_data_path}. Please check the directory.")

# 3. Start uploading to the ClearML file server
print("Uploading data to ClearML...")
dataset.upload()

# finalize() locks the dataset and makes it read-only for version control
dataset.finalize()

print("Dataset 'bdd100k_seg' uploaded and finalized successfully!")
```
