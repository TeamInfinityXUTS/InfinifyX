```python
from clearml import Dataset

# 1. 创建一个新的数据集对象
# dataset_project: ClearML 界面上的项目名称
# dataset_name: 数据集的名称
dataset = Dataset.create(
    dataset_project='InfinifyX', 
    dataset_name='bdd100k_tiny'
)

# 2. 指定你要上传的本地文件夹路径
# 按照你的描述，路径是 'InfinifyX/bdd100k_tiny'
dataset.add_files(path='bdd100k_tiny')

# 3. 开始上传到 ClearML 文件服务器
dataset.upload()

# 4. 完成并锁定数据集（锁定后数据集将变为只读，生成版本号）
dataset.finalize()

print("Dataset Upload")
```

```python
from clearml import Dataset
import os

# 1. 确认文件夹是否存在（防止路径报错）
data_path = 'bdd100k' 
if not os.path.exists(data_path):
    print(f"❌ 找不到文件夹 '{data_path}'，")
else:
    print(f"✅ 找到文件夹 '{data_path}'，准备开始创建数据集...")

  
    dataset = Dataset.create(
        dataset_project='InfinifyX', 
        dataset_name='bdd100k'
    )

    dataset.add_files(path=data_path)

    print("正在上传中，完整数据集较大，请耐心等待...")
    dataset.upload()
    dataset.finalize()

    print("🎉 完整版 bdd100k 数据集上传成功！")
```
