```python
from ultralytics import YOLO
import cv2
import os

MODEL_PATH = "runs/detect/runs/YOLOv8_BDD100K/experiment_20260323_050256/weights/last.pt"

test_images = [
    "detection_test_system_image/test1.jpg",
    "detection_test_system_image/test2.jpg",
    "detection_test_system_image/test3.jpg"
]

SAVE_DIR = "detection_result"

model = YOLO(MODEL_PATH)

for img_path in test_images:
    print(f"\n Testing: {img_path}")

    results = model(img_path)

    for r in results:
        img = r.plot()
        img_name = os.path.basename(img_path)
        save_path = os.path.join(
            SAVE_DIR,
            img_name.replace(".jpg", "_yolo_result.jpg")
        )
        cv2.imwrite(save_path, img)
        print(f" Saved: {save_path}")
```

```python
import cv2
import matplotlib.pyplot as plt

img_paths = [
    "detection_result/test1_yolo_result.jpg",
    "detection_result/test2_yolo_result.jpg",
    "detection_result/test3_yolo_result.jpg"
]

plt.figure(figsize=(15, 5))

for i, path in enumerate(img_paths):
    img = cv2.imread(path)
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    plt.subplot(1, 3, i + 1)
    plt.imshow(img)
    plt.title(f"Result {i+1}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

y_true = [0]*1292 + [1]*1346 + [2]*1362
y_pred = [
    0]*1065 + [1]*226 + [2]*1 + \
    [0]*169 + [1]*1006 + [2]*171 + \
    [0]*1 + [1]*233 + [2]*1128

cm = confusion_matrix(y_true, y_pred)
classes = ["Low", "Medium", "High"]

plt.figure(figsize=(6,5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=classes, yticklabels=classes)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Risk Prediction Model")
plt.show()
```
